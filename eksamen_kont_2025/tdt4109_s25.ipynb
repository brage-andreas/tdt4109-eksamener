# 3.1 alt_type
Funksjonen type returnerer datatypen til en variabel eller verdi. F.eks.
```python
>>> type(3)
<class 'int'>
```
Dette er bra, men nå vil vi ikke at den skal skrive ut hele den <class ' og >'-greia. Lag en ny funksjon bare_type som tar i mot en verdi eller variabel (på samme måte som vanlig type) og returnerer bare stringen "int". Eksempel på kjøring:
```python
>>> bare_type(3)
int
 
>>> x = 3.14
>>> bare_type(x)
float
```
Bruk type-funksjonen inni bare_type-funksjonen, og hent ut _bare_ datatype-delen.
 
NB: type returnerer ikke string, men du kan konvertere returverdien dens til string

````python
def alt_type(variabel):
    return str(type(variabel)).split("'")[1]

print(" 3:\t"+alt_type(3))
print("'3':\t"+alt_type('3'))
````

Gitt listen `personer` som vist under, skal vi legge til en kolonne på slutten med telefonnummer. Lag funksjonen `add_column` som legger til en kolonne til høyre for fornavn- og etternavn-kolonnene.

tlf = [91323456, 12345678, 87654321]

````python
tlf = [91323456, 12345678, 87654321]
personer = [
    ['Paul', 'Knutson'],
    ['Barack', 'Obama'],
    ['Erna', 'Solberg']
]

# Merk her, funksjonen spesifiserer at den skal ta inn personer som argument, IKKE tlf. Denne er derimot tilgjengelig siden den er spesifisert før funksjonskallet!
def add_column(liste):
    for i in range(len(liste)):
        liste[i].append(tlf[i])
    return liste

print(add_column(personer))
````

Du har fått tekstfilen "data.csv" i et CSV-liknende filformat (CSV-fil = Comma-Separeted Values) som vist her:
```
1 2 3
A B C
D E F
```
Her er separatoren mellomrom, altså er det mellomrom mellom 1, 2 og 3. Altså en Mellomrom-Separated Values-fil? MSV? Anyway.
 
Lag koden som laster den inn som en Python-tabell (2D-liste, eller liste-med-lister).
CSV-filen har headers/overskrifter (1, 2 og 3). De skal vi ikke inkludere her.
```python
>>> hent_tabell('data.csv')
[['A', 'B', 'C'], ['D', 'E', 'F']]
```
Lag funksjonen `hent_tabell` som tar inn `filnavn`, laster inn filen og returnerer den som en Python-tabell som vist i eksempelet over.


````python
def hent_tabell(filnavn):
    out = []
    with open(filnavn) as f:
        f.readline()
        for line in f.readlines():
            # strip for å fjerne linjeskift
            out.append(line.strip().split(" "))
    return out
    
````

# 3.4 øk_dato
Lag funksjonen `øk_dato` som tar i mot:
- Dato formatert som tuple: (år, måned, dag)
- Periode formattert på samme måte som dato
Disse skal legges sammen, og øk_dato skal returnere ny dato.
 
Gå ut i fra at:
- Skuddår ikke finnes
- Alle måneder har 31 dager
 
Eksempler:
```python
>>> øk_dato((2024, 11, 28), (0, 0, 1))
(2024, 11, 29)
 
>>> øk_dato((2024, 11, 28), (1, 0, 0))
(2025, 11, 28)
 
>>> øk_dato((2024, 1, 1), (1, 2, 3))
(2025, 3, 4)
 
>>> øk_dato((2024, 11, 1), (3, 2, 1))
(2028, 1, 2)
 
>>> øk_dato((2024, 10, 31), (1, 2, 3))
(2026, 1, 3)
```

````python
def øk_dato(date, period):
    y, m, d = date
    # Neste linje, definer tre variable ved
    # å hente inn en liste med tre ting inni!
    p_y, p_m, p_d = period

    d += p_d
    m += p_m
    y += p_y

    # Håndter dager > 31
    if d > 31:
        extra_m = (d - 1) // 31
        d = ((d - 1) % 31) + 1
        m += extra_m

    # Håndter måneder > 12
    if m > 12:
        extra_y = (m - 1) // 12
        m = ((m - 1) % 12) + 1
        y += extra_y

    return (y, m, d)

print(øk_dato((2024, 1, 1), (1, 2, 3)))
````

# 3.5 rec re r
Lag funksjonen `letter_count` som tar i mot:
En streng s
En dictionary d
Oppførsel og krav:
Funksjonen SKAL være rekursiv, og resultatet av kallet til funksjonen skal være:
En dictionary med tegnene i strengen s som nøkler, og antallet ganger de opptrer som verdier.
Du før ikke lov til å bruke streng-funksjonen count.
Ja, jeg vet at denne er enkel å løse uten rekursjon. Poenget med oppgaven er å se om du klarer å løse oppgaven med de verktøyene du er gitt.
 
Eksempler:
```python
>>> letter_count("banana", {})
{'b': 1, 'a': 3, 'n': 2}
```

````python
def letter_count(s, d):
    if not s:
        return d
    bokstav = s[0]
    if bokstav in d:
        d[bokstav] += 1
    else:
        d[bokstav] = 1
    return letter_count(s[1:], d)

print(letter_count("banana", {}))
````

# 3.7 - poengsamling
*data* er en éndimensjonal liste som inneholder navn og poengene de har fått i et spill, og er resultatet av forrige oppgave. Listen inneholder først navn, så dennes poeng, så neste navn og dennes poeng.

Lag funksjonen total_scores(data)
Funksjonen skal ta inn denne listen, og returnere en dictionary der nøklene er navnene på spillerne, mens verdiene inneholder skåren til hver spiller, hver nøkkel er altså en liste.

Eksempel:
```python
>>> data = ["Carl", "4", "8", "13", "Berit", "11", "15", "44", "Ada", "24", "1"]
>>> print(total_scores(data))
{'Carl': [4, 8, 13], 'Berit': [11, 15, 44], 'Ada': [24, 1]}
```

````python
def total_scores(data):
    scores = {}
    name = ""

    for value in data:
        if value.isnumeric():
            scores[name].append(int(value))
        else:
            name = value
            scores[name] = []
    return scores

scores = total_scores(data)
print(scores)
````

# 3.8 - lagre til fil
Spillernes totalskår skal nå lagres til en fil.

Lag funksjonen save_scores(scores)
Funksjonen skal ha dictionarien *scores* fra forrige oppgave som parameter, og lagre til fil slik:
- Spillernes navn og deres totalskår i filen *scores.txt*. 
- Spillerne skal sorteres etter hvor mange poeng de har fått, der høyeste skår kommer først og laveste skår kommer sist. 
- Hvis flere spillere har fått like mange poeng så skal de skrives på den samme linjen, disse spillernes navn skal i så fall sorteres etter navn (A før B) separert med komma og mellomrom.
```
>>> save_scores(scores)
```
Filen skal da ende opp med følgende innhold:
```
70: Berit
25: Ada, Carl
```

````python
def save_scores(scores:dict):
    # Vi kan lage en dictionary som lagrer navn og totalskår
    total = {}
    for k, v in scores.items():
        total[k] = sum(v)
    # Så kan vi sortere alle de unike skårene, da kan vi bruke et sett!
    sums = sorted(set(total.values()),reverse=True)
    # Åpne fil for skriving
    with open("scores.txt", "w") as f:
        # Gå igjennom sluttsummene, høyeste først
        for value in sums:
            # Samle alle navn som har fått denne skåren
            value_names = []
            for k, v in total.items():
                if v == value:
                    value_names.append(k)
            # Sette sammen strengen, join er finfin for navnene.
            out_string = str(value)+": "+", ".join(sorted(value_names))+"\n"
            f.write(out_string)

# En kortere variant som aktivt bruker dictionary og list comprehension.
# Denne er lagt med mer som et eksempel på hva en kan gjøre på få linjer!
def save_scores2(scores:dict):
    total = {k:sum(v) for k,v in scores.items()}
    with open("scores.txt", "w") as f:
        for value in sorted(set(total.values()),reverse=True):
            f.write(f"{value}: {','.join(sorted([k for k, v in total.items() if v == value]))}\n")

save_scores2(scores)
````
