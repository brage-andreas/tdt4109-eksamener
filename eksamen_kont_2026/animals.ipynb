# Dyrehagen

En dyrehage skal plassere dyr i innhegninger etter diett og størrelse. Dyrehagen har begrenset plass, og vi må sørge for at dyrene ikke spise hverandre. For at dette skal gå bra, forutsetter vi at planteetere er minst dobbelt så store som ikke-planteetere som de deler innhegning med.

## Datastruktur

Systemet baserer seg på tre hovedkomponenter:

- **species** er en ordbok der hver art har fast diett og størrelse. Dette er en global variabel og skal aldri endres.
- **animals** er en ordbok som mapper hvert individs unike navn til sin art.
- **zoo** er en 2D-liste der hver innhegning er en liste med dyrenavn.

Du har også tilgang til en hjelpefunksjon. Du trenger ikke forstå hvordan den gjør det den gjør - men bare bruk den slik den er beskrevet.

````python
ZOO_CAPACITY = 500

species = {
    "lion":    {"diet": "carnivore", "size": 50},
    "wolf":    {"diet": "carnivore", "size": 40},
    "cow":     {"diet": "herbivore", "size": 30},
    "elephant":{"diet": "herbivore", "size": 80},
    "rabbit":  {"diet": "herbivore", "size": 4},
    "raccoon": {"diet": "omnivore",  "size": 8},
    "bear":    {"diet": "omnivore",  "size": 60},
}

animals = {
    "Leo":    "lion",
    "Simba":  "lion",
    "Daisy":  "cow",
    "Bessie": "cow",
    "Bandit": "raccoon",
    "Ricky":  "raccoon",
    "Winnie": "bear",
    "Jumbo":  "elephant"
}

zoo = [
    ["Leo", "Simba"],
    ["Daisy", "Jumbo"],
]


# Hjelpefunksjon
def animals_of_diet_group(pen, animals, group):
    """Returnerer navnene på dyrene i en innhegning som tilhører en gitt diettgruppe.

    Parametre:
        pen (list[str]): liste med dyrenavn i innhegningen.
        animals (dict): ordbok som mapper navn til art.
        group (str): 'herbivore' for planteetere, 'meat_eater' for kjøttetere
                    (carnivore og omnivore).

    Returnerer:
        list[str]: navn på dyrene i innhegningen som tilhører gruppen.
    """
    result = []
    for name in pen:
        diet = species[animals[name]]["diet"]
        if group == "herbivore":
            belongs = diet == "herbivore"
        else:
            belongs = diet in ("carnivore", "omnivore")
        if belongs:
            result.append(name)
    return result
````
